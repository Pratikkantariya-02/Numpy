## Numpy array
learn about numpy 

```python
import numpy as np
```

### Creating array from list

```python
arr_id = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr_id)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D Array:\n", arr_2d)
```

### List vs numpy array

```python
py_list = [1, 2, 3]
print("Python List multiplication:", py_list * 2)

np_array = np.array([1, 2, 3])  #Element wise multiplication
print("NumPy Array multiplication:", np_array * 2)

import time
start = time.time()
py_list = [i * 2 for i in range(1000000)]
print("\n List operation time: ",time.time() - start)


start = time.time()
np_array = np.array(1000000) * 2
print("\n Numpy operation time: ",time.time() - start)
```

### Creating array from scratch

```python
zeros = np.zeros((3,4))
print("Zeros array: \n", zeros)

ones = np.ones((2,3))
print("ones array: \n",ones)

full = np.full((2,2), 7)
print("full array: \n",full)

random = np.random.random((2,3))
print("random array: \n",random)

sequence = np.arange(0,10,2)
print("sequence array: \n",sequence)
```

### Vector , Matrix and Tensor

```python
vector = np.array([1,2,3])
print("Vector: \n",vector)

matrix = np.array([[1,2,3],[4,5,6]])
print("matrix: \n",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor :\n ",tensor)
```

### Array properties

```python
arr = np.array([[1,2,3],[4,5.6,True]])
print("shape: ",arr.shape)
print("Dimension: ",arr.ndim)
print("size: ",arr.size)
print("data type: ",arr.dtype)
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original array: ",arr)

reshaped = arr.reshape((3,4))
print("\n Reshaped array: ",reshaped)

flattened = reshaped.flatten() # convert in single line
print("\n Flattened array: ",flattened)

raveled = reshaped.ravel() # its return view instead of copy
print("\n raveled array: ",raveled)

transpose = reshaped.T
print("\n Transpose array :",transpose)
```
