```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 15000, 18000, 22000, 25000], # Paradies Biryani
    [2, 12000, 14000, 16000, 19000], # Beijing Bites
    [3, 20000, 23000, 26000, 30000], # Pizza Hub
    [4, 18000, 21000, 24000, 27000], # Burger Point
    [5, 16000, 18500, 20500, 23000]  # Chai Point
])

print("=== Zomato sales analysis ===")
print("\n Sales data shape:", sales_data.shape)
print("\n Sales data for last 3 restaurants: ", sales_data[: , 1: ])
```

```python
# Total sales per year
print(np.sum(sales_data, axis = 0))
print(np.sum(sales_data[: ,1:], axis = 0))

# min and max and avg sales per restaurant
print(np.min(sales_data[: ,1:], axis=1))
print(np.max(sales_data[: ,1:], axis=0))
print(np.mean(sales_data[: ,1:], axis=1))

# cumsum
cumsum = np.cumsum(sales_data[: ,1:], axis=1)
print("C: ",cumsum)
```

```python
plt.figure(figsize=(10,6)) # define size of the graph
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumulative sales accross all restaurant ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
